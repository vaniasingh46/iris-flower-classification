# Iris Flower Classification Using Machine Learning

**Mission 1 — Beginner Machine Learning Project**

## 1. Project Idea
This project predicts the species of an Iris flower from four measurements: sepal length, sepal width, petal length, and petal width. The possible species are Iris setosa, Iris versicolor, and Iris virginica.

This is useful as a simple demonstration of how machine learning can learn patterns from labelled data and classify new observations.

## 2. Dataset
We use the classic Iris dataset included with scikit-learn. It contains 150 samples, four numeric measurement features, and a species label. Dataset documentation: https://scikit-learn.org/stable/auto_examples/datasets/plot_iris_dataset.html

## 3. Models
- **Decision Tree Classifier:** learns a sequence of feature-based decision rules.
- **K-Nearest Neighbors (KNN):** predicts a class from the labels of nearby training examples.

We use the same train-test split for both models. Feature scaling is applied to KNN because distance-based models are sensitive to feature scales.


## Step 1: Import libraries
We import tools for data handling, splitting the dataset, building models, and evaluating predictions.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print('Libraries imported successfully.')
```

## Step 2: Load the dataset
The dataset is loaded directly from scikit-learn, so no separate CSV download is needed.

```python
iris = load_iris(as_frame=True)
df = iris.frame.copy()
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))
df = df.drop(columns=['target'])

display(df.head())
print('Dataset loaded successfully.')
```

## Step 3: Inspect the data
Check the dataset size, column data types, missing values, and class distribution before training.

```python
print('Dataset shape:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nMissing values in each column:')
print(df.isnull().sum())
print('\nSpecies counts:')
print(df['species'].value_counts())
```

**Observation:** The standard Iris dataset contains 150 rows and five columns in this table (four measurements and the species label). The measurements are numeric and the dataset has no missing values.

## Step 4: Prepare features (X) and target (y)
The features are the four flower measurements. The target is the species label that the model must predict.

```python
feature_columns = iris.feature_names
X = iris.data.copy()
y = iris.target.copy()

print('Feature columns:', list(X.columns))
print('X shape:', X.shape)
print('y shape:', y.shape)
print('\nTarget classes:', list(iris.target_names))
```

## Step 5: Split into training and testing sets
We use 80% of the data for training and 20% for testing. `stratify=y` preserves the class proportions, and `random_state=42` makes the split reproducible.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print('Training samples:', len(X_train))
print('Testing samples:', len(X_test))
```

## Step 6: Build and train two models
A Decision Tree does not require feature scaling. KNN uses distances, so we put `StandardScaler` inside a pipeline. This ensures the scaler is fitted only on the training data and avoids data leakage.

```python
models = {
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'K-Nearest Neighbors (KNN)': make_pipeline(
        StandardScaler(),
        KNeighborsClassifier(n_neighbors=5)
    )
}

for name, model in models.items():
    model.fit(X_train, y_train)
    print(f'{name} trained successfully.')
```

## Step 7: Evaluate accuracy
Accuracy is the proportion of test examples predicted correctly. Since this is a small dataset, the scores can vary with the train-test split; treat them as a learning exercise rather than a guarantee of real-world performance.

```python
results = []
predictions = {}

for name, model in models.items():
    y_pred = model.predict(X_test)
    predictions[name] = y_pred
    accuracy = accuracy_score(y_test, y_pred)
    results.append({'Model': name, 'Accuracy': accuracy, 'Accuracy (%)': round(accuracy * 100, 2)})

results_df = pd.DataFrame(results).sort_values('Accuracy', ascending=False).reset_index(drop=True)
display(results_df)

best_model_name = results_df.loc[0, 'Model']
print(f'Best model on this test split: {best_model_name}')
```

## Step 8: Display confusion matrices
A confusion matrix compares the actual species with the predicted species. Correct predictions appear on the main diagonal; off-diagonal values indicate misclassifications.

```python
class_names = list(iris.target_names)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
for ax, (name, y_pred) in zip(axes, predictions.items()):
    cm = confusion_matrix(y_test, y_pred, labels=[0, 1, 2])
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
                xticklabels=class_names, yticklabels=class_names, ax=ax)
    ax.set_title(name)
    ax.set_xlabel('Predicted species')
    ax.set_ylabel('Actual species')
plt.tight_layout()
plt.show()
```

## Step 9: Classification reports
Precision, recall, and F1-score provide more detail than accuracy alone.

```python
for name, y_pred in predictions.items():
    print(f'--- {name} ---')
    print(classification_report(y_test, y_pred, target_names=class_names, zero_division=0))
```
